## SEARCH

```python
s1 = "huynh trung nghia"
pattern = 'ng'
result = re.search(pattern,s1)
print(type(result))
if result: print('Match found!')
else: print('Match not found!')
```

## MATCH

```python
print(re.match("huynh", s1))
print(re.match("trung", s1))
```

## GROUP - dung chung voi search/match
group(0) - toan bo 
group(1) - nhom 1
group(2) - nhom 2

```python
import re
text = "Name: Nghia, Age: 19"
pattern = r"Name: (\w+), Age: (\d+)"
result = re.search(pattern, text)
print(result)
print(result.group(1))
print(result.group(2))
```

## FINDALL

```python
pattern = r"\W"
example = r"solvar/huynh "
matches = re.findall(pattern,example)
print(f"Matches: {matches}")
```

```python
s2 = "The BodyGuard is the best album of 'Whitney Houston'."


# Use the findall() function to find all occurrences of the "st" in the string
result = re.findall("st", s2)

# Print out the list of matched words
print(type(result))
```

## SPLIT

```python
split_list = re.split(r"\s", s2)
print(split_list)
```

## SUB

```python
pattern = "whitNEY hOUSton"
replacement = "Trung Nghia"
result = re.sub(pattern, replacement, s2, flags = re.IGNORECASE)
print(result)
print(s2) # khong sua chuoi ban dau
```
