## Scrapy

```python
import scrapy
from scrapy.crawler import CrawlerProcess

class QuotesScrapy(scrapy.Spider):
    name = 'quotes'
    start_urls = ['http://quotes.toscrape.com/tag/humor/',]
    def parse(self, response):
        for quote in response.css('div.quote'):
            # yield {'quote': quote.css('span.text::text').get()}
            text = quote.css('span.text::text').get()
            
            # Viết thêm dòng này để in ra màn hình cho dễ thấy
            print(f"\n ĐÃ LẤY ĐƯỢC: {text} \n") 
            
            yield {'quote': text}

# process = CrawlerProcess()

# process.crawl(QuotesScrapy)

# process.start()
```

## Web Scrapy with Pandas

```python
import pandas as pd
url = 'https://en.wikipedia.org/wiki/List_of_largest_banks'
tables = pd.read_html(url,storage_options={'User-Agent': 'Mozilla/5.0'}) #list - moi phan tu la 1 dataframe
df = tables[0]
print(df)
```
