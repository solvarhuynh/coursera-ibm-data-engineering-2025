```python
from pycoingecko import CoinGeckoAPI
```

```python
cg = CoinGeckoAPI() #tạo 1 object kết nói với máy chủ (như ng đại diện để đưa ra request)

price = cg.get_price(ids='bitcoin', vs_currencies='usd') #gửi 1 request (get_price) cho cg để cg gửi tới máy chủ
# sau đó máy chủ gửi lại két quả (data - dictionary)

price
```

```python
bitcoin_data = cg.get_coin_market_chart_by_id(id = 'bitcoin', vs_currency = 'usd', days=30)
```

```python
bitcoin_data
```

```python
bitcoin_data['prices']
```

```python
import pandas as pd
data = pd.DataFrame(bitcoin_data['prices'], columns = ['TimeStamp','Price'])
data
```

```python
data['Date'] = pd.to_datetime(data['TimeStamp'], unit = 'ms')
data
```

```python
candlestick_data = data.groupby(data.Date.dt.date).agg({'Price': ['min','max','first','last']})
candlestick_data
```

```python
import plotly.graph_objects as go
fig = go.Figure(data=[go.Candlestick(x = candlestick_data.index,
        open = candlestick_data['Price']['first'],
        high = candlestick_data['Price']['max'],
        low = candlestick_data['Price']['min'],
        close = candlestick_data['Price']['last']
        )]
    )
fig.update_layout(xaxis_rangeslider_visible = False,
    xaxis_title = 'Date',
    yaxis_title = 'Price (USD $)',
    title = 'Bitcoin Candlestick Chart Over Past 30 Days')
```
