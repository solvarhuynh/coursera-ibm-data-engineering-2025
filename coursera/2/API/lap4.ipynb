```python
from randomuser import RandomUser
import pandas as pd
```

```python
r = RandomUser()
```

## GENERATE_USERS

```python
some_list = r.generate_users(10)
```

```python
some_list
```

```python
name = r.get_full_name()
name
```

```python
for user in some_list:
    print(user.get_full_name(), user.get_email(), sep = ':  ')
```

```python
def get_user():
    users = []
    for user in some_list:
        users.append({"FUll Name" : user.get_full_name(), "Age" : user.get_age(), "Gender" : user.get_gender(), "City" : user.get_city(), "State" : user.get_state(), "Email" : user.get_email(), "Dob" : user.get_dob(), "Picture" : user.get_picture()})
    return pd.DataFrame(users)
```

```python
import os
import requests
import json
```

```python
url_line = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/data/Example1.txt"
r_line = requests.get(url_line)
path = os.path.join(os.getcwd(),'three_line')
with open(path, 'w') as f:
    f.write(r_line.text)
with open(path, 'r') as f:
    print(f.read())
```

```python
url_fruit = "https://web.archive.org/web/20240929211114/https://fruityvice.com/api/fruit/all"
data = requests.get(url_fruit)
data_fruit = json.loads(data.text) #list (dict )
df_data_fruit = pd.DataFrame(data_fruit)
df_data_fruit
```

```python
df2 = pd.json_normalize(data_fruit)
df2
```
